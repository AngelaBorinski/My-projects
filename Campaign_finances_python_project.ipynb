```python
with open('CandidateSummaryAction1.csv', 'r') as csv_file:
  data = csv.reader(csv_file)
  net_con = []

  for line in data:
    net_con.append(line[43])
#above code creates a list with all the net contribution values for each candidate

  net_con.pop(0) #removes first item in the list, which is "net_con" and not a number
  #print(net_con[0:10])- prints first 10 values in the list. Used full print statements to check an entire list, or only the first 10 if I wanted to see that my list edits worked without having to scroll a bunch

  net_con = [s.replace("$", "").replace(" ", "").replace(",", "").replace("(", "").replace(")", "") for s in net_con]
  while("" in net_con):
    net_con.remove("")
  #above code removed blank values and non-numerical characters in the list, which I cannot do calculations on

  #print(net_con[0:10])

  # for item in net_con:
  #   try:
  #     print(float(item))
  #   except ValueError:
  #        print(f"Cannot convert '{item}' to float.")
  # used the above code when I couldn't convert the net contributions to floats to see which items in the list were causing the error


  for index in range(len(net_con)):
    net_con[index] = float(net_con[index])
  #print(net_con[0:10])
  #converts to floats and prints out first ten items in the list

  #first calculation: find the average net contribution
  total = 0
  for x in net_con:
    total+=x
  avg = total/len(net_con)
  rounded_avg = round(avg, 2)
  print("The average net contribution for candidates in this dataset is $" + str(rounded_avg))
```

```python
with open('CandidateSummaryAction1.csv', 'r') as csv_file:
  data = csv.reader(csv_file)

  votes = []
  for line in data:
    votes.append(line[50])
    #creates a list of the number of votes candidates in the database received

  #print(votes)
  votes.pop(0)
  #print(votes)

  # for item in votes:
  #   try:
  #     print(int(item))
  #   except ValueError:
  #     print(f"Cannot convert '{item}' to integer.")
  # also used to check to see where to convert

  while("" in votes):
    votes.remove("")

  for index in range(len(votes)):
    votes[index] = int(votes[index])
  #print(votes)

#calculates what percentage of candidates had more than 200 thousand votes
  over_200k = []
  for x in votes:
    if x > 200000:
      over_200k.append(x)
    else:
      pass
  #print(len(over_200k))
  #print(len(votes))
  division = len(over_200k)/len(votes)
  percentage = division*100
  rounded_percentage = round(percentage, 2)
  print(str(rounded_percentage) + "% of candidates in the database had more than 200k votes")
```
